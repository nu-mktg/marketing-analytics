````python
# Setup verification — run this cell first.
# It imports EVERY library this course uses. Anything marked ❌ here would
# otherwise have failed silently in the week you first needed it.

import importlib
import sys

import numpy as np       # noqa: F401  (bound here so later cells can use np/pd)
import pandas as pd      # noqa: F401

print(f"Python {sys.version.split()[0]}   (this course is pinned to 3.11)\n")

LIBRARIES = [
    ("numpy",           "numpy",        "arrays and linear algebra"),
    ("pandas",          "pandas",       "dataframes"),
    ("scipy",           "scipy",        "distributions and optimisation"),
    ("matplotlib",      "matplotlib",   "plotting"),
    ("seaborn",         "seaborn",      "statistical plots"),
    ("statsmodels",     "statsmodels",  "regression, OLS, MMM"),
    ("scikit-learn",    "sklearn",      "segmentation, classification"),
    ("scikit-uplift",   "sklift",       "uplift modelling (HW07)"),
    ("lifetimes",       "lifetimes",    "customer lifetime value (HW08)"),
    ("lifelines",       "lifelines",    "survival analysis (HW04)"),
    ("prophet",         "prophet",      "demand forecasting (HW10)"),
    ("cmdstanpy",       "cmdstanpy",    "the Bayesian backend prophet uses"),
    ("pycausalimpact",  "causalimpact", "CausalImpact (HW14)"),
]

failed = []
for name, module, why in LIBRARIES:
    try:
        mod = importlib.import_module(module)
        print(f"  ✅ {name:<16}{getattr(mod, '__version__', '—'):<10}{why}")
    except Exception as exc:
        failed.append((name, exc))
        print(f"  ❌ {name:<16}{'FAILED':<10}{why}")

print()
if failed:
    print(f"⚠️  {len(failed)} of {len(LIBRARIES)} libraries did not import.\n")

    # A NEW CODESPACE REPORTS ITSELF READY BEFORE IT IS — measured 2026-09-02 in
    # a real Codespace built from the course template: it reached the "Available"
    # state, and the editor opened, while the background install was still
    # running with only 63 of ~100 packages present and prophet missing
    # entirely. So the FIRST question on an import failure is not "what is
    # broken" but "has it finished installing", because the two need opposite
    # actions — wait, versus rebuild — and a student who rebuilds a
    # perfectly-good Codespace just starts the same wait over again.
    still_installing = False
    try:
        import subprocess as _sp
        _r = _sp.run(["pgrep", "-af", "pip install"], capture_output=True,
                     text=True, timeout=5)
        still_installing = "requirements" in (_r.stdout or "")
    except Exception:
        pass

    if still_installing:
        print("   ⏳ YOUR ENVIRONMENT IS STILL INSTALLING. Nothing is wrong.")
        print("      Wait two or three minutes and re-run this cell. Do not")
        print("      rebuild, and do not pip install anything.")
    else:
        print("   If you created this Codespace in the last few minutes, it may")
        print("   still be installing in the background even though the editor")
        print("   looks ready — wait a couple of minutes and re-run this cell")
        print("   before assuming anything is broken.\n")
        print("   If it still fails after that, rebuild the Codespace")
        print("   (Codespaces → ... → Rebuild container) rather than pip")
        print("   installing, so your numbers still match what the grader")
        print("   expects. If it STILL fails, post the output below to Canvas.")
    for name, exc in failed:
        print(f"     {name}: {type(exc).__name__}: {exc}")
else:
    print(f"✅ Environment check passed — all {len(LIBRARIES)} course libraries")
    print("   import at their pinned versions. You are ready to work.")
    print("\n   (prophet prints 'Importing plotly failed' above — that is")
    print("    expected. plotly is not used in this course.)")

# ── setup_report.json ─────────────────────────────────────────────────────────
# Writes a small file next to this notebook recording ONLY what is printed
# above: your Python version, the version of each library, which ones failed,
# and whether this ran in a Codespace. It contains nothing personal. Commit it
# along with your work — it is how the course confirms your environment is
# correct without needing to look over your shoulder, and it is what lets us
# help you before a broken library costs you a homework.
#
# It is written as a FILE rather than left in the cell output on purpose: output
# is easy to clear by accident, and a file survives that. It is also parsed as
# JSON rather than by reading the text above, so rewording a message here can
# never quietly break the check that reads it.
import json as _json
import os as _os
import datetime as _dt

try:
    _report = {
        "python": sys.version.split()[0],
        "libraries": {name: getattr(sys.modules.get(module), "__version__", None)
                      for name, module, _ in LIBRARIES if module in sys.modules},
        "failed": [name for name, _ in failed],
        "in_codespace": _os.environ.get("CODESPACES") == "true",
        "written_at": _dt.datetime.now(_dt.timezone.utc).isoformat(timespec="seconds"),
    }
    # A plain relative path, deliberately. The notebook's working directory IS
    # its own folder (the devcontainer sets jupyter.notebookFileRoot to
    # ${fileDirname}), so this lands in hw00/ where the course looks for it.
    # This was os.path.dirname(os.path.abspath("__file__")) until 2026-09-03 —
    # "__file__" in QUOTES is a string literal, not the variable, so it resolved
    # to the working directory and the file landed at the repo ROOT. Caught by a
    # live student-account run, not by reading the code.
    with open("setup_report.json", "w") as _fh:
        _json.dump(_report, _fh, indent=1)
    print("\n📄 Wrote setup_report.json — commit it with your work.")
except Exception as _exc:
    print(f"\n(could not write setup_report.json: {_exc} — not a problem, "
          "continue with the homework)")
````
